```python
import pandas as pd
from pathlib import Path

# ---------------------------------------------------------
# Load known circles
# ---------------------------------------------------------

EGO_ID = "100129275726588145876"

circle_file = Path(
    f"data/raw/gplus/{EGO_ID}.circles"
)

circles = {}

with open(circle_file, "r", encoding="utf-8") as f:

    for line in f:

        parts = line.strip().split()

        if len(parts) >= 2:

            circle_id = parts[0]
            members = set(parts[1:])

            circles[circle_id] = members


# ---------------------------------------------------------
# Convert Louvain partition into communities
# ---------------------------------------------------------

louvain_communities = {}

for node, community_id in partition.items():

    if community_id not in louvain_communities:
        louvain_communities[community_id] = set()

    louvain_communities[community_id].add(node)


# ---------------------------------------------------------
# Compare communities with circles
# ---------------------------------------------------------

results = []

for community_id, community_nodes in louvain_communities.items():

    for circle_id, circle_nodes in circles.items():

        intersection = (
            community_nodes &
            circle_nodes
        )

        union = (
            community_nodes |
            circle_nodes
        )

        if union:

            jaccard = (
                len(intersection) /
                len(union)
            )

        else:

            jaccard = 0

        circle_coverage = (
            len(intersection) /
            len(circle_nodes)
        )

        community_precision = (
            len(intersection) /
            len(community_nodes)
        )

        results.append({
            "community_id": community_id,
            "circle_id": circle_id,
            "community_size": len(community_nodes),
            "circle_size": len(circle_nodes),
            "overlap": len(intersection),
            "circle_coverage": circle_coverage,
            "community_precision": community_precision,
            "jaccard": jaccard
        })


comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    "jaccard",
    ascending=False
)

comparison_df.head(20)
```
